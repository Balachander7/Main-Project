```python
# Drop email column  (Mail ID's are not use ful for Organizational level Analytics)

df = df.drop(columns=["Mail_ID"],errors='ignore')
```

```python
# Strip extra spaces from column header names

df.columns = df.columns.str.strip()
```

```python
df.duplicated().sum()   # No Duplicates available
```

 Standardize  formatting

```python
# Employee name Column standardization

df['Employee_Name'] = df['Employee_Name'].astype(str).str.strip().str.replace(r'\s+', ' ', regex=True).str.title()
```

```python
 # Gender Column Standardization

df['Gender']= df["Gender"].astype(str).str.strip().str.upper()   ## Strip leading/trailing whitespaces and convert text to uppercase
df['Gender'] = df["Gender"].replace({'M':'Male','MALE':'Male','F':'Female','FEMALE':'Female','NAN': np.nan})

print(df['Gender'].value_counts())
```

```python
#Department Column Standardization

df['Department'] = df['Department'].str.strip().str.upper()
df['Department']=df['Department'].replace({
    'TECHNOLOGY': 'Technology', 'IT': 'Technology',
    'SALES': 'Sales', 'SALES & MKTG': 'Sales',
    'OPERATIONS': 'Operations', 'OPS': 'Operations',
    'HUMAN RESOURCES': 'Human Resources', 'HR': 'Human Resources', 'H.R.': 'Human Resources',
    'FINANCE': 'Finance',
    'ENGINEERING': 'Engineering',
    'MARKETING': 'Marketing','NAN':np.nan
})

print(df['Department'].value_counts())
```

```python
# Designation Column Standardization

df['Designation'] = df['Designation'].str.strip().str.upper()
df['Designation']=df['Designation'].replace({
    'EXECUTIVE': 'Executive',
    'ASSISTANT MANAGER': 'Assistant Manager',
    'MANAGER': 'Manager',
    'SENIOR EXECUTIVE': 'Senior Executive',
    'TEAM LEAD': 'Team Lead','NAN':np.nan,'NP.NAN': np.nan
})

print(df['Designation'].value_counts())
```

```python
# Region / Location Standardization

df['Region_Location'] = df['Region_Location'].str.strip().str.upper()
df['Region_Location']=df['Region_Location'].replace({
    'BENGALURU': 'South - Bangalore', 'BANGALORE': 'South - Bangalore', 'SOUTH - BANGALORE': 'South - Bangalore',
    'CHENNAI': 'South - Chennai', 'SOUTH - CHENNAI': 'South - Chennai',
    'DELHI': 'North - Delhi', 'NORTH - DELHI': 'North - Delhi',
    'MUMBAI': 'West - Mumbai', 'WEST - MUMBAI': 'West - Mumbai',
    'KOLKATA': 'East - Kolkata', 'EAST - KOLKATA': 'East - Kolkata',"NAN":np.nan
})

print(df['Region_Location'].value_counts())
```

```python
# Source of Hiring Column Standardization


df['Source_Of_Hiring'] = df['Source_Of_Hiring'].str.strip().str.upper()
df['Source_Of_Hiring']=df['Source_Of_Hiring'].replace({
    'NAUKRI': 'Naukri',
    'LINKEDIN': 'LinkedIn',
    'COMPANY CAREER SITE': 'Company Career Site',
    'EMPLOYEE REFERRAL': 'Employee Referral',
    'RECRUITMENT AGENCY': 'Recruitment Agency',
    'WALK-IN': 'Walk-In',
    'CAMPUS DRIVE': 'Campus Drive','NAN': np.nan
})

print(df['Source_Of_Hiring'].value_counts())
```

```python
# Experience_Type Column Standardization

df['Experience_Type'] = df['Experience_Type'].str.strip().str.upper()
df['Experience_Type']=df['Experience_Type'].replace({
    'FRESHER': 'Fresher',
    'EXPERIENCED': 'Experienced','NAN':np.nan
})

print(df['Experience_Type'].value_counts())
```

```python
#Employment Status Column Standardization

df['Employment_Status'] = df['Employment_Status'].str.strip().str.upper()
df['Employment_Status']=df['Employment_Status'].replace({
    'ACTIVE': 'Active',
    'INACTIVE': 'Inactive',
    'RESIGNED': 'Resigned',
    'TERMINATED': 'Terminated','NAN':np.nan
})

print(df['Employment_Status'].value_counts())
```

```python
# Work Mode Column Standardization

df['Work_Mode'] = df['Work_Mode'].str.strip().str.upper()
df['Work_Mode']=df['Work_Mode'].replace({
    'WFH': 'Work From Home', 'WORK FROM HOME': 'Work From Home',
    'ONSITE': 'Onsite',
    'HYBRID': 'Hybrid','NAN':np.nan
})


print(df['Work_Mode'].value_counts())
```

```python
# Transportation Type Column Standardization

df['Transportation_Type'] = df['Transportation_Type'].str.strip().str.upper()
df['Transportation_Type']=df['Transportation_Type'].replace({
    'OWN VEHICLE': 'Own Vehicle',
    'COMPANY CAB': 'Company Cab',
    'PUBLIC TRANSPORT': 'Public Transport','NAN': np.nan
})

print(df['Transportation_Type'].value_counts())
```

```python
df['Flight_Risk_Indicator'] = df['Flight_Risk_Indicator'].astype(str).str.strip().str.capitalize().replace({'Nan': np.nan})

print(df['Flight_Risk_Indicator'].value_counts())
```

```python

# Clean numeric columns stored as text

numeric_Columns = ['Cost_Per_Hire', 'Annual_CTC', 'Productivity_Score', 'KPI_Achievement_Rate',
    'Engagement_Score', 'Satisfaction_Score', 'Sick_Leaves_Taken',
    'Late_Arrival_Count', 'Training_Hours', 'Overtime_Hours', 'Performance_Rating',
    'Days_Present', 'Casual_Leaves_Taken', 'Unplanned_Absences']  # Create a new list with the all numeric columns for easy access


for col in numeric_Columns:
  clean_series = df[col].astype(str).str.replace('%','').str.replace(',','').str.replace('$','').str.strip()
  df[col]=pd.to_numeric(clean_series,errors='coerce')

```

```python
# 8. Convert date text into datetime objects

date_cols = ['Date_Of_Birth', 'Date_Of_Joining', 'Last_Working_Date']  # Create a new list with all the Date columns for easy access

for col in date_cols:
  df[col]=pd.to_datetime(df[col],format= 'mixed',errors='coerce')
```

Handling Missing Values (Imputation Strategy)

```python
# 1. Imputation for Exit Reason column
df['Reason_For_Leaving'] = df['Reason_For_Leaving'].fillna('Not Applicable(Active Employee)') # For active employees Reason for leaving not required.
```

```python
# 2. Impute Numeric Columns with Median

for col in numeric_Columns:
  df[col]= df[col].fillna(df[col].median())
```

```python
# 3. Impute Categorical Columns with Mode

categorical_cols = ['Gender', 'Department', 'Designation', 'Region_Location', 'Source_Of_Hiring',
                    'Experience_Type', 'Work_Mode', 'Transportation_Type', 'Flight_Risk_Indicator', 'Employment_Status']

for col in categorical_cols:
  df[col]=df[col].fillna(df[col].mode()[0])

```

```python
# 4. Handle the missing values in Managers Emp code

df['Manager_Employee_ID']=df['Manager_Employee_ID'].fillna("DIR-0001")  # Assing the Directer's code as Reporitng manager code Where its null

```

```python
# Handle the Missing values in Date columns

df['Date_Of_Birth'] = df['Date_Of_Birth'].fillna(df['Date_Of_Birth'].mode()[0]) ## Mode
df['Date_Of_Joining'] = df['Date_Of_Joining'].ffill()  ## Forward Fill
```

Outlier Treatment (IQR Capping Method)

```python
for col in ['Annual_CTC', 'Cost_Per_Hire']:
  for col in ['Annual_CTC', 'Cost_Per_Hire']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = max(0, Q1 - (1.5 * IQR))
    upper_bound = Q3 + (1.5 * IQR)



  if lower_bound < upper_bound:
        df[col] = np.clip(df[col], lower_bound, upper_bound)

```

New Column Creation

```python
# 1. Calculate Employee Age

df['Age']= 2026- df['Date_Of_Birth'].dt.year
```

```python
# 2. Calculate Total Absences
df['Total_Absences'] = df['Sick_Leaves_Taken'] + df['Casual_Leaves_Taken'] + df['Unplanned_Absences']
```

```python
# 3. Calculate Tenure in Years
end_dates = df['Last_Working_Date'].fillna(pd.Timestamp('2026-09-20'))
df['Tenure_Years'] = ((end_dates - df['Date_Of_Joining']).dt.days / 365.25).round(1)

df['Tenure_Years'] = df['Tenure_Years'].apply(lambda x: max(0, x))
```

```python
# Average Cost Per Hire by Recruitment Source (Bar Plot)

plt.figure(figsize=(8, 5))
source_df = df.groupby('Source_Of_Hiring')['Cost_Per_Hire'].mean().reset_index()
ax = sns.barplot(data=source_df, x='Source_Of_Hiring', y='Cost_Per_Hire', palette='Blues_d')

plt.title('1. Sourcing Cost Efficiency by Hiring Channel', fontsize=12, fontweight='bold')
plt.xlabel('Source of Hiring', fontsize=10)
plt.ylabel('Average Cost Per Hire (₹)', fontsize=10)
plt.xticks(rotation=20)

for p in ax.patches:
    if p.get_height() >= 0:
        ax.annotate(f'₹{p.get_height():.0f}',
                    (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='center', xytext=(0, 5), textcoords='offset points', fontweight='bold')

plt.tight_layout()
plt.show()
```

### **Insights**

* point by point insight should be recorded

## Interpretation of your above chart

Point 1 (Before Cleaning): The raw dataset contained messy text in Cost_Per_Hire with prefixes like 'INR', commas (e.g., '78,465'), and missing values. The Source_Of_Hiring column had inconsistent casing like 'naukri', 'LinkedIn', and 'COMPANY CAREER SITE'.

Point 2 (After Cleaning): Text prefixes and commas were stripped out, converting Cost_Per_Hire into clean numerical float values. Hiring channel names were standardized across all records.

Point 3 (Business Insight):

 * Finding: Campus Drives (₹6,090) and Recruitment Agencies (₹6,034) incur the highest hiring costs, whereas organic channels like LinkedIn, Naukri, and Company Career Sites cost ₹0.

 * Recommendation: Reallocate budget from third-party agencies toward an internal Employee Referral Program (paying a ₹2,500 bonus per successful hire).

 * Business Impact: Reduces overall recruitment spending by ~40% while encouraging high-quality candidate referrals.

```python
# Overtime Hours Impact on KPI Achievement Rate (Bar Plot)

df['Overtime_Group'] = pd.cut(df['Overtime_Hours'], bins=[-1, 10, 20, 30, 100], labels=['0-10 hrs', '11-20 hrs', '21-30 hrs', '30+ hrs'])
ot_kpi = df.groupby('Overtime_Group')['KPI_Achievement_Rate'].mean().reset_index()

plt.figure(figsize=(8, 5))
ax = sns.barplot(data=ot_kpi, x='Overtime_Group', y='KPI_Achievement_Rate', palette='Purples_d')

plt.title('3. Overtime Hours Impact on KPI Achievement Rate', fontsize=12, fontweight='bold')
plt.xlabel('Overtime Hours per Month', fontsize=10)
plt.ylabel('Average KPI Achievement Rate (%)', fontsize=10)

for p in ax.patches:
    if p.get_height() > 0:
        ax.annotate(f'{p.get_height():.1f}%',
                    (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='center', xytext=(0, 5), textcoords='offset points', fontweight='bold')

plt.tight_layout()
plt.show()
```

## Interpretation of your above chart

Point 1 (Before Cleaning): The KPI_Achievement_Rate column contained percentage strings with '%' symbols, decimals like 0.76, and missing values.  

Point 2 (After Cleaning): Special symbols were stripped, and values were normalized onto a uniform 0–100 percentage scale.

Point 3 (Business Insight & Action Plan):

Finding: Moderate overtime (11–20 hours/month) yields the highest KPI achievement (66.1%). Excessive overtime (30+ hours/month) leads to fatigue and diminishing returns, dropping KPI achievement to 63.7%.

Recommendation: Cap maximum monthly overtime at 20 hours per employee and trigger automated managerial alerts when exceeded.   Business Impact: Saves unnecessary overtime payout costs while optimizing workforce performance and accuracy.   

```python
# Manager Level Performance Metrics
mgr_df = df.groupby('Manager_Employee_ID').agg(
    Emp_Count=('Employee_ID', 'count'),
    High_Risk_Count=('Flight_Risk_Indicator', lambda x: (x == 'High').sum()),
    Avg_Productivity=('Productivity_Score', 'mean')
).reset_index()

mgr_df['Risk_Rate_%'] = (mgr_df['High_Risk_Count'] / mgr_df['Emp_Count']) * 100
top_mgrs = mgr_df[mgr_df['Emp_Count'] >= 20].copy()

plt.figure(figsize=(8,6))
sns.scatterplot(
    data=top_mgrs,
    x='Avg_Productivity',
    y='Risk_Rate_%',
    size='Emp_Count',
    sizes=(100, 700),
    hue='Risk_Rate_%',
    palette='Reds',
    alpha=0.7,
    legend=False
)

for idx, row in top_mgrs.iterrows():
    plt.text(row['Avg_Productivity'], row['Risk_Rate_%'], row['Manager_Employee_ID'], fontsize=9, fontweight='bold', ha='center')

plt.title('1. Manager Effectiveness Matrix: Productivity vs. Flight Risk % (Bubble Chart)', fontsize=12, fontweight='bold')
plt.xlabel('Average Productivity Score (0-100)', fontsize=10)
plt.ylabel('Flight Risk Percentage (%)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

The raw data had 85 missing Manager IDs and unformatted employee records, making it impossible to see team sizes or track how individual managers were performing.

2) After Cleaning:
We filled the missing manager IDs, grouped employees by their reporting manager, and calculated each team's total headcount, average productivity score, and high attrition (flight) risk percentage.

3) Business Insight & Action Plan:

  Finding: Managers DIR-0001 (195 staff) and EMP-0050 (87 staff) have dangerously high team attrition risks (63.6% and 64.4%), whereas manager EMP-0200 keeps risk low (43.5%) while maintaining top productivity (61.2%).

  Recommendation: Provide leadership coaching and mandatory weekly 1-on-1s for high-risk managers, using EMP-0200’s team management practices as a model.
